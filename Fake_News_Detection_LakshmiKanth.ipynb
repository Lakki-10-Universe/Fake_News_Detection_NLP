#### 2.1.1 Write the function to clean the text and remove all the unnecessary elements  <font color = red>[4 marks]</font> <br>

```python
# Write the function here to clean the text and remove all the unnecessary elements
import string

def clean_text(column):
    # Convert to lower case
    column = column.str.lower()

    # Remove text in square brackets
    column = column.replace(r'\[.*?\]', '', regex=True)
    
    # Remove punctuation
    punctuation_to_remove = string.punctuation
    column = column.apply(lambda x:x.translate(str.maketrans("","",string.punctuation)))

    # Remove words with numbers
    column = column.replace(r'\b\w*\d\w*\b', '', regex=True)
    
    return column
```

#### 2.1.2  Apply the function to clean the news text and store the cleaned text in a new column within the new DataFrame. <font color = red>[1 mark]</font> <br>

```python
# Creating another small function to check if there are any other weird characters in the text.
import re

def find_weird_characters(df, column_name):

    weird_texts = df[df[column_name].apply(lambda x: bool(re.search(r'[^A-Za-z0-9\s\.,;:\'\"!?()-]', str(x))))]
    
    all_text = ' '.join(df[column_name].dropna())
    weird_chars = re.findall(r'[^A-Za-z0-9\s\.,;:\'\"!?()-]', all_text)
    unique_weird_chars = set(weird_chars)
    
    return weird_texts, unique_weird_chars
```

### **2.2** POS Tagging and Lemmatization  <font color = red>[10 marks]</font> <br>



#### 2.2.1 Write the function for POS tagging and lemmatization, filtering stopwords and keeping only NN and NNS tags <font color = red>[8 marks]</font> <br>



```python
# Write the function for POS tagging and lemmatization, filtering stopwords and keeping only NN and NNS tags
def lemma_pos_tag(column):
    
    doc = nlp(column)
    lemma_pos = [(token.lemma_, token.pos_) for token in doc if not token.is_stop and token.tag_ in {'NN', 'NNS'}]
    return lemma_pos
```

#### 2.2.2  Apply the POS tagging and lemmatization function to cleaned text and store it in a new column within the new DataFrame. <font color = red>[2 mark]</font> <br>

**NOTE: Store the cleaned text and the lemmatized text with POS tags removed in separate columns within the new DataFrame.**

**This will be useful for analysing character length differences between cleaned text and lemmatized text with POS tags removed during EDA.**


```python
# Apply POS tagging and lemmatization function to cleaned text
# Store it in a separate column in the new DataFrame
df_clean['lemma_pos'] = df_clean['cleaned_news_text'].progress_apply(lambda x : lemma_pos_tag(x))
```

```python
# creating a column for lemmatized text without POS tags for further analysis 
df_clean['lemma_text'] = df_clean['lemma_pos'].progress_apply(lambda x : " ".join([lemma for lemma, pos in x]))
```

### **4.3** Find and display the top unigrams, bigrams and trigrams by frequency in true news and fake news after processing the text  <font color = red>[20 marks]</font> <br>




##### 4.3.1 Write a function to get the specified top n-grams  <font color = red>[4 marks]</font> <br>



```python
# Write a function to get the specified top n-grams

def generate_ngrams(text, n, top_n):
    # Cleaning and filtering words using lambda
    words = list(filter(lambda words: words.isalpha() and len(words) > 2, text.split()))
    
    # Creating n-grams as tuples
    ngrams = list(map(lambda i: tuple(words[i:i+n]), range(len(words) - n + 1)))
    
    # Return top n most frequent n-grams
    return Counter(ngrams).most_common(top_n)
```

```python
# We will also create another small funtion to plot the bar graphs to represent the ngrams

def plot_ngram_graph(ngram_data, title="Top N-grams", palette="viridis"):
    # Create a DataFrame from the n-gram data
    df = pd.DataFrame(ngram_data, columns=["ngram", "count"])
    df["ngram"] = df["ngram"].apply(lambda x: " ".join(x))

    plt.figure(figsize=(10, 5))
    ax = sns.barplot(
        data=df.sort_values("count", ascending=False), 
        y="ngram", 
        x="count", 
        palette=palette
    )

    for p in ax.patches:
        count = int(p.get_width())
        percent = 100 * count / df["count"].sum()
        label = f'{count} ({percent:.0f}%)'
        ax.annotate(
            label, 
            (p.get_width(), p.get_y() + p.get_height() / 2),
            ha="left", 
            va="center",
            fontsize=10,
            color="black",
            fontweight="bold",
            xytext=(5, 0),
            textcoords="offset points"
        )

    plt.title(title, fontsize=14, fontweight='bold')
    plt.xlabel("Frequency", fontsize=12)
    plt.ylabel("N-grams", fontsize=12)
    #plt.grid(axis='x', linestyle='--', alpha=0.7)
    plt.tight_layout()
    plt.show()
```

##### 7.1.2 Calculate and print accuracy, precision, recall and f1-score on validation data <font color = red>[5 marks]</font>

```python
# Writing a small function to calculate all the necessary metrics for models and the Classification Report instead of repeating everytime.
def evaluate_model(y_train, y_train_pred, y_test, y_test_pred, model_name="Model"):
    metrics = {
        "Accuracy": [accuracy_score(y_train, y_train_pred), accuracy_score(y_test, y_test_pred)],
        "Precision": [precision_score(y_train, y_train_pred), precision_score(y_test, y_test_pred)],
        "Recall": [recall_score(y_train, y_train_pred), recall_score(y_test, y_test_pred)],
        "F1 Score": [f1_score(y_train, y_train_pred), f1_score(y_test, y_test_pred)]
    }
    
    print(f"\nMetrics for {model_name}:\n" + "-"*50)
    for metric, (train_val, test_val) in metrics.items():
        print(f"{metric:10} - Train: {train_val:.4f} | Test: {test_val:.4f}")
    print("-"*50)

# For Classification Report
def show_classification_report(y_train, y_train_pred, y_test, y_test_pred, model_name="Model"):
    print(f"\nClassification Report for {model_name}\n" + "="*50)
    print("\nTrain Data Report:\n")
    print(classification_report(y_train, y_train_pred))
    
    print("\nTest Data Report:\n")
    print(classification_report(y_test, y_test_pred))
    print("="*50)
```

```python
## Calculate and print accuracy, precision, recall, f1-score on predicted labels

evaluate_model(y_train, y_pred_lgr_train, y_test, y_pred_lgr_test, model_name="Logistic Regression")
```
